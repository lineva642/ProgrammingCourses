```python
df_pyspark = spark.read.csv('pyspark_practice_2.csv', sep = ';', header=True, inferSchema=True)
```

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols=['Age', 'Experience', 'Salary'],
    outputCols=["{}_imputed".format(c) for c in ['Age', 'Experience', 'Salary']]
).setStrategy("median")
```

```python
# Add imputation cols to df
imputer.fit(df_pyspark).transform(df_pyspark).show()
```

## GroupBY & Aggregate Functions


```python
df_pyspark = spark.read.csv('pyspark_practice_3.csv', sep = ';', header=True, inferSchema=True)
```

```python
df_pyspark.show()
```

```python
df_pyspark.printSchema()
```

```python
df_pyspark = df_pyspark.drop('Age')
```

```python
## Groupby
df_pyspark.select(['Name', 'Salary', 'Department']).groupBy('Name').sum().show()
```

```python
df_pyspark.groupBy('Department').sum().show()
```

```python
df_pyspark.groupBy('Department').mean().show()
```

```python
df_pyspark.groupBy('Department').count().show()
```

```python
df_pyspark.agg({'Salary': 'sum'}).show()
```

```python
df_pyspark.groupBy('Name').avg().show()
```
